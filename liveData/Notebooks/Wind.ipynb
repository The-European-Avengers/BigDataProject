```python
# Import libraries
import requests
import pandas as pd
import folium
import geopandas as gpd
from shapely.geometry import Point
```

```python
# API Parameters
api_key = "9c1728d6-69ec-4be2-8e6c-72dee4dc44df"  
url = "https://dmigw.govcloud.dk/v1/forecastedr/collections/harmonie_dini_sf/cube"

params = {
    "bbox": "7.0,54.5,16.0,58.0",        # All denmark
    "parameter-name": "wind-speed-10m",
    "datetime": "2025-10-17T00:00:00Z/..",
    "crs": "crs84",
    "f": "GeoJSON",
    "api-key": api_key
}
```

```python
# Call the API
response = requests.get(url, params=params)
if response.status_code != 200:
    raise Exception(f"API request failed: {response.status_code}, {response.text}")

data = response.json()
print("Data retrieved successfully. Number of points:", len(data['features']))
```

```python
# Convert GeoJSON a DataFrame
records = []
for feature in data['features']:
    lon, lat = feature['geometry']['coordinates']
    props = feature['properties']
    props.update({"lon": lon, "lat": lat})
    records.append(props)

df = pd.DataFrame(records)
print("DataFrame created. Rows:", df.shape[0])
```

```python
# Show first response value
print(data['features'][:1])
```

```python
# Show data values as a table
features = data['features']

properties = [
    {
        # Extract coordinates from geometry
        'lat': feature['geometry']['coordinates'][1],
        'lon': feature['geometry']['coordinates'][0],
        # Merge all properties
        **feature['properties']
    }
    for feature in features
]

df = pd.DataFrame(properties)
df.head()
```

```python
# Show the data for one point
# Convert 'step' column to datetime if it’s not already
df['step'] = pd.to_datetime(df['step'])

# Sort by latitude, longitude, then time
df_sorted = df.sort_values(by=['lat', 'lon', 'step'])

df_sorted.head(10)  # first 10 rows
```

```python
# Load shapefile,  shapefile shoulb be in the folder: .shp, .shx, .dbf, .prj)
dk_shape = gpd.read_file("../dk.shp")  
dk_shape = dk_shape.to_crs("EPSG:4326")  # Make sure is CRS WGS84
```

```python
# Convert DataFrame to GeoDataFrame
gdf = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df.lon, df.lat), crs="EPSG:4326")
```

```python

# Filter points, just the ones inside Denmark
gdf = gdf[gdf.within(dk_shape.unary_union)]
print("Points Inside Denmark:", gdf.shape[0])
```

```python
# Select a timestamp to visualize
timestep = gdf['step'].unique()[0]  # primer timestep
gdf_step = gdf[gdf['step'] == timestep]
```

```python
# Create map with Folium
import folium
from IPython.display import display

m = folium.Map(location=[55.5, 10.0], zoom_start=6)

for _, row in gdf_step.iterrows():
    speed = row['wind-speed-10m']
    folium.CircleMarker(
        location=[row['lat'], row['lon']],
        radius=4,
        color="blue",
        fill=True,
        fill_color="blue",
        fill_opacity=0.7,
        popup=f"Wind: {speed:.2f} m/s\nTime: {row['step']}"
    ).add_to(m)

# Show map inline
display(m)
```
